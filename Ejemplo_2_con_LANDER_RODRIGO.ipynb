**Objetivo:** abrir un archivo CSV, observar sus datos.

Usaremos `productos_ventas.csv`, separado por punto y coma (`;`).

## 1. Abrir el archivo con Pandas
Antes, debemos subir el archivo a Google Colab (a la izquierda clic en la carpeta **Archivos** y elegir el icono de flecha que es para **Subir al almacenamiento de sesión.**

`pd.read_csv()` lee el CSV. El parámetro `sep=';'` indica el separador de columnas.

```python
import pandas as pd


df = pd.read_csv('productos_ventas.csv', sep=';')
display(df.head(10))  #muestra primeros 20 registros
df.tail(2)  #muestra últimos 2 registros
```

## 2. Saber cuántas filas y columnas tiene

```python
print('Filas y columnas:', df.shape)
print('Filas:', len(df))
print('Filas:', df.shape[0])
print('Columnas:', df.shape[1])
```

## 3. Mostrar los nombres de las columnas

```python
print(df.columns.tolist())
```

## 4 Saber los tipos de datos de cada columna

```python
print(df.dtypes)
```

## 5. Consultar una columna

Por ejemplo, mostrar solamente los nombres de los productos.

```python
print(df['producto'])
```

## 6. Filtrar productos por precio

Mostrar los productos cuyo precio es mayor que 500.000.

```python
display(df[df['precio'] > 500000])
```

## 7. Calcular un promedio

Calcularemos el precio promedio (con dos decimales) de todos los registros.

```python
print('Precio promedio:', round(df['precio'].mean(), 2))
```

## 8. Ejemplos estadísticos con NumPy

Calculamos el precio máximo y el mínimo.

```python
import numpy as np

precioMax = df['precio'].max()
print('Precio máximo:', precioMax)
precioMin = df['precio'].min()
print('Precio mínimo:', precioMin)
```

## 9. Resumen de datos

Agruparemos por la categoría

```python
# Agrupar por categoría y calcular el promedio de stock
df_agrupado = df.groupby('categoria')['stock'].mean().round(0)
display(df_agrupado)
df_agrupado = df_agrupado.astype(int)
display(df_agrupado)
```
